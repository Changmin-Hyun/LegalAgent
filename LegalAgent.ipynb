```python
# 2. LLM 초기화
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
# 3. 시스템 프롬프트 엔지니어링 (페르소나 및 가이드라인 부여)
system_prompt = """
당신은 일반인들이 일상에서 겪는 상황을 법률적으로 해석하고 이해하기 쉽게 설명해주는 '법률 가이드 에이전트'입니다.

[작동 원칙]
1. 사용자의 상황을 분석하여 관련된 주요 법적 쟁점(예: 민사상 손해배상, 근로기준법 위반 등)을 파악합니다.
2. 전문적인 법률 용어는 최소화하고, 불가피하게 사용할 경우 중학생 수준의 어휘로 풀어서 설명합니다.
3. 구체적인 승패 여부를 단정 짓지 말고, '현재 상황에서 주장할 수 있는 권리'와 '취할 수 있는 일반적인 법적 절차(예: 내용증명 발송, 노동청 신고 등)'를 객관적으로 제시합니다.
4. 마지막 줄에는 반드시 다음 문구를 정확히 포함하세요:
   "⚠️ 본 답변은 일반적인 법률 상식에 기반한 참고용이며, 실제 법적 조치를 취하기 전에는 반드시 변호사, 노무사 등 전문가와 상담하시기 바랍니다."
"""
```

```python
# 프롬프트 템플릿 구성
prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", "현재 상황: {situation}")
])
```

```python
# 4. LCEL(LangChain Expression Language)을 활용한 체인(Chain) 구성
legal_agent_chain = prompt | llm | StrOutputParser()
```

```python
# 5. 실행 함수
def analyze_situation(situation_text: str) -> str:
    print("에이전트가 상황을 분석 중입니다...\n" + "-"*50)
    # chain.invoke를 통해 프롬프트에 변수를 주입하고 실행
    response = legal_agent_chain.invoke({"situation": situation_text})
    return response
```

```python
# === 테스트 실행 ===
if __name__ == "__main__":
    # 일반인이 겪을 수 있는 임의의 상황
    sample_situation = "두 달 전 헬스장 1년 치를 끊었는데, 갑자기 이사를 가게 되어서 환불을 요청했습니다. 그런데 헬스장 측에서는 특가 할인으로 계약했기 때문에 위약금 30%를 내야 한다고 합니다. 이게 맞는 건가요?"
    
    result = analyze_situation(sample_situation)
    print(result)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langgraph.prebuilt import create_react_agent
from langchain_community.tools import DuckDuckGoSearchRun
from langgraph.checkpoint.memory import MemorySaver # 💡 메모리 기능 추가
```

```python
# 1. 툴(Tools) 및 스킬(Skills) 정의

web_search = DuckDuckGoSearchRun(
    name="web_search",
    description="최신 법률 상식, 판례, 뉴스, 법령 정보 등을 실시간으로 인터넷에서 검색할 때 반드시 사용하세요."
)

@tool
def analyze_legal_context(text: str, analysis_type: str) -> str:
    """
    주어진 텍스트(text)를 분석하여 법적 쟁점을 파악합니다.
    분석 목적에 따라 analysis_type을 다음 중 하나로 지정하세요:
    - 'situation': 억울한 일, 층간소음, 분쟁, 범죄 피해 등의 상황일 때 관련 법령 및 용어 검색
    - 'contract': 계약서나 합의서 문구일 때 불리한 독소조항 분석
    """
    print(f"\n[시스템] 통합 법률 분석 툴 작동 중... (타입: {analysis_type})")
    result = []
    
    # 1. 상황 분석 로직
    if analysis_type == 'situation':
        if "개" in text and "짖" in text:
            result.append("[상황 분석] 관련 법률: '경범죄 처벌법 제3조(인근소란 등)', '민법 제750조 불법행위로 인한 손해배상(위자료)'")
        else:
            result.append("[상황 분석] 관련 법률: '정보통신망법 위반(명예훼손)', '형법 제311조 모욕죄' 등")
            
    # 2. 계약서 분석 로직
    elif analysis_type == 'contract':
        warnings = []
        if "위약금" in text:
            warnings.append("- [주의] 과도한 위약금 약정은 민법상 무효가 될 소지가 있습니다.")
        if "포기" in text:
            warnings.append("- [위험] 사전에 법적 권리(예: 퇴직금, 소송제기권)를 포기하게 만드는 조항은 효력이 없을 수 있습니다.")
        
        if warnings:
            result.append("[계약서 검토 결과] 다음 독소조항이 의심됩니다:\n" + "\n".join(warnings))
        else:
            result.append("[계약서 검토 결과] 텍스트 내 명백한 독소조항 키워드는 없습니다.")
            
    else:
        return "지원하지 않는 분석 타입입니다. 'situation' 또는 'contract'를 사용하세요."

    return "\n\n".join(result)

@tool
def calculate_legal_interest(principal: int, rate: float, months: int) -> str:
    """
    원금(principal), 연 이자율(rate, %), 지연 기간(months, 개월)을 입력받아 지연 이자를 계산합니다.
    금전적 피해 보상이나 체불 임금 이자를 물어볼 때 사용하세요.
    """
    interest = principal * (rate / 100) * (months / 12)
    total = principal + interest
    return f"지연 이자는 {int(interest):,}원이며, 원금 포함 총 수령액은 {int(total):,}원입니다."


@tool
def write_legal_draft(doc_type: str, sender: str, recipient: str, summary: str) -> str:
    """
    사용자가 내용증명, 합의서 등 법률 문서 작성을 요청할 때 이 툴을 사용해 초안을 작성합니다.
    """
    print(f"\n[시스템] {doc_type} 초안 작성 스킬 가동 중...")
    
    draft = f"""
    [ {doc_type} 초안 ]
    
    ■ 발신인: {sender}
    ■ 수신인: {recipient}
    ■ 제  목: 계약 해지 및 환불 요청의 건
    
    1. 귀사(하)의 무궁한 발전을 기원합니다.
    2. 주요 내용: {summary}
    3. 본 서면을 수령하신 후 7일 이내에 원만한 합의(환불)가 이루어지지 않을 경우, 
       소비자원 피해구제 신청 및 민사상 법적 조치를 취할 예정임을 알려드립니다.
       
    2026년 OO월 OO일
    발신인: {sender} (인)
    """
    return draft
```

```python
# 2. LLM 및 툴 리스트 설정
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
tools = [web_search, analyze_legal_context, calculate_legal_interest, write_legal_draft]
```

```python
system_prompt = """
당신은 일반인들이 일상에서 겪는 상황을 돕는 '법률 가이드 에이전트'입니다.
1. 전문 용어는 중학생 수준으로 풀어서 설명하세요.
2. 초안 작성을 요청받으면 반드시 'write_legal_draft' 툴을 사용하세요.
3. 툴이 반환한 초안 내용은 절대로 요약하거나 생략하지 말고, 원본 텍스트 전체를 100% 그대로 답변에 포함해서 출력하세요.
4. 마지막 줄에는 "⚠️ 본 답변은 참고용이며, 실제 조치 전 전문가와 상담하세요."를 포함하세요.
"""
```

```python
# 3. LangGraph를 이용한 에이전트 생성
memory = MemorySaver()
legal_agent = create_react_agent(llm, tools, prompt=system_prompt, checkpointer=memory, interrupt_before=["tools"])
```

```python
def chat_with_agent(user_input: str, thread_id: str) -> str:
    config = {"configurable": {"thread_id": thread_id}}
    
    # 1. 사용자의 질문을 넣고 에이전트 실행 시작
    response = legal_agent.invoke(
        {"messages": [("user", user_input)]}, 
        config=config
    )
    
    # 💡 2. 에이전트가 완전히 종료될 때까지 무한 반복하며 상태 확인
    while True:
        state = legal_agent.get_state(config)
        
        # 더 이상 진행할 작업(next)이 없으면 대화 완성! 루프 탈출
        if not state.next:
            break
            
        # 에이전트가 툴 사용 직전에 멈췄다면
        if state.next[0] == "tools":
            last_message = state.values["messages"][-1]
            tool_calls = getattr(last_message, 'tool_calls', [])
            
            # 초안 작성(write_legal_draft) 툴이 포함되어 있는지 확인
            needs_approval = any(tc["name"] == "write_legal_draft" for tc in tool_calls)
            
            if needs_approval:
                print("\n✋ [승인 대기] 에이전트가 초안을 작성하려고 합니다.")
                user_approval = input("진행할까요? (Y/N): ")
                
                if user_approval.strip().upper() == "Y":
                    print("[시스템] 승인 완료. 실행을 이어갑니다...\n")
                    response = legal_agent.invoke(None, config=config) # 멈춘 곳부터 다시 시작
                else:
                    return "사용자에 의해 작업이 취소되었습니다."
            else:
                # 검색 등 승인이 필요 없는 툴은 묻지 않고 자동으로 이어서 실행
                response = legal_agent.invoke(None, config=config)

    # 3. 모든 과정이 끝나면 최종 메시지 추출 (기존과 동일)
    messages = legal_agent.get_state(config).values["messages"]
    final_message = messages[-1].content
    if isinstance(final_message, list):
        final_message = final_message[0].get('text', str(final_message))
        
    tool_draft = ""
    for msg in reversed(messages):
        if msg.type == "human": 
            break
        if msg.type == "tool" and "초안" in msg.content:
            tool_draft = msg.content
            break
            
    if tool_draft and tool_draft not in final_message:
        return f"{tool_draft}\n\n[에이전트 추가 코멘트]\n{final_message}"
        
    return final_message
```

```python
# 테스트 1 실행

if __name__ == "__main__":
    session_id = "user_773"

    sample_situation = """
    이웃집 개가 밤마다 너무 짖어서 수면 장애가 왔어. 법적으로 어떻게 대응할 수 있을까?
    """
    
    result = chat_with_agent(sample_situation,session_id)
    print("\n[최종 답변]\n")
    print(result)
```
